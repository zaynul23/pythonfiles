```python
from datetime import datetime
import json
import logging
from typing import Any, Dict
from bson import ObjectId

client = MongoClient("mongodb://localhost:27017")
db = client["linguallink"]
print(db.list_collection_names())
roomcol=db["rooms"]
messages_collection=db["messages"]
# logger = logging.getLogger("backend.messages")


def _sanitize_bson(obj: Any) -> Any: # dict to bson
    # Recursively sanitize BSON types
    if isinstance(obj, dict):
        return {k: _sanitize_bson(v) for k, v in obj.items()}
    if isinstance(obj, list):
        return [_sanitize_bson(v) for v in obj]
    
    # handle actual ObjectId instances explicitly
    if isinstance(obj, ObjectId):
        return str(obj)
    
    if isinstance(obj, datetime):
        return obj.isoformat()
    
    return obj


async def save_message(doc: Dict[str, Any]) -> Dict[str, Any]:
    """Save a message document to Mongo and return a sanitized document ready for broadcast.

    The `doc` should include keys: room_id, text, sender_id, sender_email, sender_language, translations, created_at.
    """
    if "created_at" not in doc:
        doc["created_at"] = datetime.utcnow()

    res = await messages_collection.insert([doc])
    doc["id"] = str(res.inserted_id)
    # Convert created_at to ISO
    if isinstance(doc.get("created_at"), datetime):
        doc["created_at"] = doc["created_at"].isoformat()

    return _sanitize_bson(doc)


def dumps_safe(obj: Any) -> str:
    return json.dumps(obj, default=str)

```

```python
from pymongo import MongoClient
from elasticsearch import Elasticsearch, helpers

MONGO_URI = "mongodb://localhost:27017"
ES_URI = "http://localhost:9200"

# -------------------------------
# 1. CONNECT TO MONGO
# -------------------------------
mongo = MongoClient(MONGO_URI)
db = mongo["testdb"]
collection = db["EStest"]

# -------------------------------
# 2. CONNECT TO ELASTICSEARCH
# -------------------------------
es = Elasticsearch(
    "http://localhost:9200",
    basic_auth=("elastic", "lunyaz")
)
index_name = "estest"   # same as Mongo collection name

# -------------------------------
# 3. CREATE INDEX WITH MAPPING
# -------------------------------

mapping = {
    "mappings": {
        "properties": {
            "message_id":   {"type": "keyword"},
            "room_id":      {"type": "keyword"},

            "sender_id":    {"type": "integer"},
            "sender_email": {"type": "keyword"},
            "sender_name":  {"type": "keyword"},
            "sender_language": {"type": "keyword"},

            "en": {"type": "text", "analyzer": "english"},
            "es": {"type": "text", "analyzer": "spanish"},
            "fr": {"type": "text", "analyzer": "french"},
            "hi": {"type": "text", "analyzer": "standard"},
            "ur": {"type": "text", "analyzer": "standard"},

            "created_at": {"type": "date"}
        }
    }
}

# Delete old index if exists (optional)
if es.indices.exists(index=index_name):
    es.indices.delete(index=index_name)

# Create new index
es.indices.create(index=index_name, body=mapping)


# -------------------------------
# 4. PREPARE BULK INDEX DATA
# -------------------------------

actions = []

for doc in collection.find({}):
    # Remove Mongo _id because ES cannot serialize ObjectId
    if "_id" in doc:
        doc.pop("_id")

    action = {
        "_index": index_name,
        "_id": doc["message_id"],   # use message_id as ES document ID
        "_source": doc
    }
    actions.append(action)

# -------------------------------
# 5. BULK INSERT INTO ES
# -------------------------------
helpers.bulk(es, actions)

print(f"Indexed {len(actions)} documents into Elasticsearch index '{index_name}'")
```

```python

from elasticsearch import Elasticsearch

LANG_FIELD = {
    "en": "en",
    "es": "es",
    "fr": "fr",
    "hi": "hi",
    "ur": "ur"
}

def search_messages(
    es: Elasticsearch,
    index_name: str,
    room_id: str,
    query_text: str,
    query_lang: str,
    sender_id: int = None,
    start_date: str = None,
    end_date: str = None,
    page_size: int = 20,
    search_after_token=None
    ):

    # Pick correct language column to search
    field = LANG_FIELD[query_lang]

    # ----------------------
    # BUILD ES QUERY
    # ----------------------
    body = {
        "size": page_size,
        "query": {
            "bool": {
                "must": [
                    {   # always restrict to room_id
                        "term": {
                            "room_id": room_id
                        }
                    },
                    {   # search only the correct language column
                        "match": {
                            field: {
                                "query": query_text,
                                "fuzziness": "AUTO"
                            }
                        }
                    }
                ],
                "filter": []
            }
        },
        "sort": [
            {"created_at": "desc"},
            {"message_id": "asc"}
        ]
    }

    # pagination
    if search_after_token:
        body["search_after"] = search_after_token

    # optional sender filter
    if sender_id is not None:
        body["query"]["bool"]["filter"].append({
            "term": {"sender_id": sender_id}
        })

    # optional date filters
    if start_date or end_date:
        drange = {"range": {"created_at": {}}}
        if start_date:
            drange["range"]["created_at"]["gte"] = start_date
        if end_date:
            drange["range"]["created_at"]["lte"] = end_date
        body["query"]["bool"]["filter"].append(drange)

    # ----------------------
    # EXECUTE SEARCH
    # ----------------------
    res = es.search(index=index_name, body=body)
    hits = res["hits"]["hits"]

    # ----------------------
    # CLEAN RESULTS
    # ----------------------
    results = []
    for h in hits:
        src = h["_source"]
        results.append({
            "message_id": src["message_id"],
            "room_id": src["room_id"],
            "sender_id": src["sender_id"],
            "sender_name": src["sender_name"],

            # return only the column we searched
            "text": src[field],

            "created_at": src["created_at"],
            "sort_token": h["sort"]   # needed for next page
        })

    next_token = hits[-1]["sort"] if hits else None

    return results, next_token

```
