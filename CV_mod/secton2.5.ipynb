```python
import cv2
import numpy as np

# 1. Prepare your input image and histogram
# Assuming you have an image 'img' and a histogram 'hist'

# 2. Convert the image to a floating-point image (required for calcBackProject)
img_float = img.astype(np.float32)

# 3. Calculate the back projection
backproj = cv2.calcBackProject([img_float], [0], histimg, [0, 256], scale=1)

# 4. Threshold the back projection image (optional, but often helpful)
_, thresh = cv2.threshold(backproj, 50, 255, cv2.THRESH_BINARY)

# 5. Display the results (optional)
cv2.imshow("Back Projection", backproj)
cv2.imshow("Thresholded Back Projection", thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

cap = cv2.VideoCapture(0)

# Read the first frame and initialize the background model
ret, frame = cap.read()
avg = np.float32(frame)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Accumulate the weighted average
    cv2.accumulateWeighted(frame, avg, 0.01)

    # Convert the avg image to uint8
    background = cv2.convertScaleAbs(avg)

    # Display the result
    cv2.imshow('Live Feed', frame)
    cv2.imshow('Background Model', background)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(0)

ret, frame1 = cap.read()
ret, frame2 = cap.read()

while True:
    # Compute the absolute difference between two frames
    diff = cv2.absdiff(frame1, frame2)

    # Convert to grayscale and threshold to highlight motion areas
    gray = cv2.cvtColor(diff, cv2.COLOR_BGR2GRAY)
    _, thresh = cv2.threshold(gray, 25, 255, cv2.THRESH_BINARY)

    # Show the thresholded difference
    cv2.imshow("Motion", thresh)

    # Read the next frame
    frame1 = frame2
    ret, frame2 = cap.read()
    if not ret:
        break

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
